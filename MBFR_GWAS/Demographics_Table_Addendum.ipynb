## Chi Squared for the Binary Traits

```python

# Input counts and totals
data = {
    'Trait': ['Diabetes', 'Obesity', 'HCL', 'Hypertension',
              'Chronic Kidney Disease', 'Coronary Artery Disease',
              'Congestive Heart Failure', 'Peripheral Artery Disease'],
    'AFR_Female': [229, 40, 276, 323, 148, 150, 135, 67],
    'AFR_Male': [165, 36, 215, 231, 159, 119, 108, 47],
    'EUR_Female': [84, 15, 141, 139, 66, 81, 62, 38],
    'EUR_Male': [171, 20, 260, 265, 159, 183, 111, 75]}

totals = {
    'AFR_Female': 335,
    'AFR_Male': 238,
    'EUR_Female': 152,
    'EUR_Male': 277}
df = pd.DataFrame(data)

# Add percentages
for group in ['AFR_Female', 'AFR_Male', 'EUR_Female', 'EUR_Male']:
    print(df[group])
    print(totals[group])
    print((df[group] / totals[group]) * 100)
    df[f'{group}_Pct'] = (df[group] / totals[group]) * 100

# Calculate totals
df['AFR_Total'] = df['AFR_Female'] + df['AFR_Male']
df['EUR_Total'] = df['EUR_Female'] + df['EUR_Male']
df['AFR_Total_Pct'] = (df['AFR_Total'] / (totals['AFR_Female'] + totals['AFR_Male'])) * 100
df['EUR_Total_Pct'] = (df['EUR_Total'] / (totals['EUR_Female'] + totals['EUR_Male'])) * 100

# Chi-squared tests - sex differences within ancestries
p_sex_afr = []
p_sex_eur = []
p_ancestry = []

for _, row in df.iterrows():
    # AFR sex comparison
    afr_table = [[row['AFR_Female'], totals['AFR_Female'] - row['AFR_Female']],
                 [row['AFR_Male'], totals['AFR_Male'] - row['AFR_Male']]]
    _, p1, _, _ = chi2_contingency(afr_table)

    # EUR sex comparison
    eur_table = [[row['EUR_Female'], totals['EUR_Female'] - row['EUR_Female']],
                 [row['EUR_Male'], totals['EUR_Male'] - row['EUR_Male']]]
    _, p2, _, _ = chi2_contingency(eur_table)

    # AFR vs EUR comparison (total counts)
    ancestry_table = [[row['AFR_Total'], (totals['AFR_Female'] + totals['AFR_Male']) - row['AFR_Total']],
                      [row['EUR_Total'], (totals['EUR_Female'] + totals['EUR_Male']) - row['EUR_Total']]]
    _, p3, _, _ = chi2_contingency(ancestry_table)

    p_sex_afr.append(p1)
    p_sex_eur.append(p2)
    p_ancestry.append(p3)

def format_p(p):
    if p < 0.001:
        return "<0.001"
    else:
        return f"{p:.3f}"
df['p_Sex_AFR'] = p_sex_afr
df['p_Sex_AFR'] = df['p_Sex_AFR'].apply(format_p)

df['p_Sex_EUR'] = p_sex_eur
df['p_Sex_EUR'] = df['p_Sex_EUR'].apply(format_p)

df['p_AFR_vs_EUR'] = p_ancestry
df['p_AFR_vs_EUR'] =df['p_AFR_vs_EUR'].apply(format_p)

df_rounded = df.round(3)
df_rounded[['Trait', 'AFR_Female', 'AFR_Female_Pct', 'AFR_Male', 'AFR_Male_Pct', 'AFR_Total', 'AFR_Total_Pct',
            'EUR_Female', 'EUR_Female_Pct', 'EUR_Male', 'EUR_Male_Pct', 'EUR_Total', 'EUR_Total_Pct',
            'p_Sex_AFR', 'p_Sex_EUR', 'p_AFR_vs_EUR']]
```

## Welch's T-Test for Quantitative Table

```python
import pandas as pd
import numpy as np
from scipy.stats import t

# Sample sizes from previous count table
n_afr_female = 335
n_afr_male = 238
n_eur_female = 152
n_eur_male = 277

def welch_ttest(mean1, std1, n1, mean2, std2, n2):
    se1 = std1**2 / n1
    se2 = std2**2 / n2
    t_stat = (mean1 - mean2) / np.sqrt(se1 + se2)
    df = (se1 + se2)**2 / ((se1**2 / (n1 - 1)) + (se2**2 / (n2 - 1)))
    p_value = 2 * t.sf(np.abs(t_stat), df)
    return p_value

# Quantitative traits data: mean and std
traits_data = {
    'Trait': ['Exam Age', 'BMI', 'A1C', 'Total Cholesterol', 'LDL Cholesterol'],
    'AFR Female': [(58.52, 10.39), (38.48, 9.07), (7.47, 1.98), (193.52, 95.47), (117.46, 94.81)],
    'AFR Male': [(57.38, 12.42), (34.42, 8.92), (7.66, 2.14), (163.32, 43.90), (89.30, 41.00)],
    'EUR Female': [(62.57, 12.74), (32.99, 9.17), (6.99, 1.63), (175.96, 50.16), (87.77, 39.09)],
    'EUR Male': [(64.00, 11.38), (31.77, 6.49), (6.94, 1.49), (163.97, 65.96), (85.46, 67.05)],
}

# Calculate p-values
pvals_within_afr = []
pvals_within_eur = []
pvals_between_total = []

for trait in traits_data['Trait']:
    afr_mean_std_female = dict(traits_data)['AFR Female'][traits_data['Trait'].index(trait)]
    afr_mean_std_male = dict(traits_data)['AFR Male'][traits_data['Trait'].index(trait)]
    eur_mean_std_female = dict(traits_data)['EUR Female'][traits_data['Trait'].index(trait)]
    eur_mean_std_male = dict(traits_data)['EUR Male'][traits_data['Trait'].index(trait)]
    
    # AFR Female vs Male
    p_afr = welch_ttest(afr_mean_std_female[0], afr_mean_std_female[1], n_afr_female,
                        afr_mean_std_male[0], afr_mean_std_male[1], n_afr_male)
    
    # EUR Female vs Male
    p_eur = welch_ttest(eur_mean_std_female[0], eur_mean_std_female[1], n_eur_female,
                        eur_mean_std_male[0], eur_mean_std_male[1], n_eur_male)
    
    # Total AFR vs EUR (using combined means/SDs
    afr_total_mean = (afr_mean_std_female[0] * n_afr_female + afr_mean_std_male[0] * n_afr_male) / (n_afr_female + n_afr_male)
    eur_total_mean = (eur_mean_std_female[0] * n_eur_female + eur_mean_std_male[0] * n_eur_male) / (n_eur_female + n_eur_male)
    
    afr_total_std = np.sqrt(
        ((n_afr_female - 1) * afr_mean_std_female[1]**2 + (n_afr_male - 1) * afr_mean_std_male[1]**2) / (n_afr_female + n_afr_male - 2)
    )
    eur_total_std = np.sqrt(
        ((n_eur_female - 1) * eur_mean_std_female[1]**2 + (n_eur_male - 1) * eur_mean_std_male[1]**2) / (n_eur_female + n_eur_male - 2)
    )
    
    p_total = welch_ttest(afr_total_mean, afr_total_std, n_afr_female + n_afr_male,
                          eur_total_mean, eur_total_std, n_eur_female + n_eur_male)
    
    pvals_within_afr.append(p_afr)
    pvals_within_eur.append(p_eur)
    pvals_between_total.append(p_total)

# Combine into dataframe
results = pd.DataFrame({
    'Trait': traits_data['Trait'],
    'AFR Female vs Male (p)': [f"{p:.3g}" if p >= 0.001 else "<0.001" for p in pvals_within_afr],
    'EUR Female vs Male (p)': [f"{p:.3g}" if p >= 0.001 else "<0.001" for p in pvals_within_eur],
    'AFR vs EUR Total (p)': [f"{p:.3g}" if p >= 0.001 else "<0.001" for p in pvals_between_total]
})

results
```
